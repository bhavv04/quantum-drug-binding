```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All imports successful")
```

```python
url = "https://deepchemdata.s3-us-west-1.amazonaws.com/datasets/bace.csv"
df = pd.read_csv(url)

# We only need these three columns
df = df[['mol', 'CID', 'pIC50']]

print(f"Dataset shape: {df.shape}")
print(df.head())
print(f"\nColumns: {df.columns.tolist()}")
```

```python
plt.figure(figsize=(8, 4))
plt.hist(df['pIC50'], bins=30, color='#534AB7', edgecolor='white')
plt.xlabel('pIC50 (binding affinity)')
plt.ylabel('Number of molecules')
plt.title('Distribution of binding affinity — BACE-1 inhibitors')
plt.tight_layout()
plt.show()

print(f"Total molecules: {len(df)}")
print(f"Mean pIC50:      {df['pIC50'].mean():.2f}")
print(f"Min pIC50:       {df['pIC50'].min():.2f}")
print(f"Max pIC50:       {df['pIC50'].max():.2f}")
print(f"Missing values:  {df['pIC50'].isna().sum()}")
```

```python
df.to_csv("bace_clean.csv", index=False)
print("Saved to bace_clean.csv")
```
